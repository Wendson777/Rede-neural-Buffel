```python
# Importação das bibliotecas
from ultralytics import YOLO
import cv2
import numpy as np

# ======= CONFIGURAÇÕES =======
            # Imagem contendo as sementes (6g)
PESO_MEDIO_SEMENTE = 0.00302               # Exemplo: 0.002g por semente (ajuste conforme sua média)
PESO_TOTAL_AMOSTRA = 0.152               # Peso total da amostra (em gramas)

# ======= 1. Carregamento do modelo YOLOv11 =======
model = YOLO('/content/drive/MyDrive/train/weights/best.pt')

# ======= 2. Realização da inferência (detecção) =======
results = model('/content/drive/MyDrive/Colab Notebooks/Teste_P.jpg')

# ======= 3. Contagem das sementes detectadas =======
contagem_sementes = len(results[0].boxes)

# ======= 4. Estimativa do peso total das sementes puras =======
peso_estimado_puro = contagem_sementes * PESO_MEDIO_SEMENTE

# ======= 5. Cálculo do percentual de pureza física =======
percentual_pureza = (peso_estimado_puro / PESO_TOTAL_AMOSTRA) * 100

# ======= 6. Exibição dos resultados =======
print(f"Quantidade de sementes detectadas: {contagem_sementes}")
print(f"Peso estimado das sementes puras: {peso_estimado_puro:.3f}g")
print(f"Percentual de pureza física: {percentual_pureza:.2f}%")

# # ======= 7. (Opcional) Exibir a imagem com as detecções =======
results[0].plot(show=True)
```
